```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
from tensorflow.keras.datasets import mnist
import warnings
warnings.filterwarnings("ignore")
```

```python
#Load datasets 
(X_train,Y_train),(X_test,Y_test)=mnist.load_data()
print("shape of X train: ",X_train.shape)
print("shape of Y train: ",Y_train.shape)
print("shape of X test: ",X_test.shape)
print("shape of Y test: ",Y_test.shape)
```

```python
#Normalize data (hamesha features ko hi normalize kiya jata h)
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
#create the model(ek ke baad ek layer bane isliye sequential ka use kiye)
model=Sequential([
    Flatten(input_shape=(28,28)), #28 pixel length and 28 pixel ht h (flatten 28*28 img ko 1D array m change karta h)
    #hidden layer 1
    Dense(128,activation='relu'), #relu is rectified linear unit(agar 0 hoga to 0 ayega & agar +ve ho to +ve dega)
    #hidden layer 2
    Dense(64,activation='relu'),
    #output layer
    Dense(10,activation='softmax')
])
#softmax converts the output scores into probabilities.
```

```python
#before training,tensorflow needs three things:
#optimizer:-how to update the weights.
#loss function:-how to calculate the error.
#metrices:-how to measure the model's performance.
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
#train the model
model.fit(X_train,Y_train,epochs=10,validation_data=(X_test,Y_test))
```

```python
#save the model
model.save("ann_model.keras")
print("Model is saved")
```
